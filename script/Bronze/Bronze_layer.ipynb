# Bronze Layer: load raw CSV files as tables
**Rule of Bronze:** copy the data as it is. No cleaning here.

## Settings
One list describes all files. To add a new file, just add one more line.

```python
CATALOG  = "workspace"
RAW_PATH = "/Volumes/workspace/bronze/source_systems"

INGESTION_CONFIG = [
    # source folder, file name,         bronze table name
    {"source": "crm", "file": "cust_info.csv",     "table": "crm_cust_info"},
    {"source": "crm", "file": "prd_info.csv",      "table": "crm_prd_info"},
    {"source": "crm", "file": "sales_details.csv", "table": "crm_sales_details"},
    {"source": "erp", "file": "CUST_AZ12.csv",     "table": "erp_cust_az12"},
    {"source": "erp", "file": "LOC_A101.csv",      "table": "erp_loc_a101"},
    {"source": "erp", "file": "PX_CAT_G1V2.csv",   "table": "erp_px_cat_g1v2"},
]
```

## Load each file into a Bronze table

```python
for item in INGESTION_CONFIG:
    path  = f"{RAW_PATH}/source_{item['source']}/{item['file']}"
    table = f"{CATALOG}.bronze.{item['table']}"

    df = (
        spark.read
             .option("header", "true")
             .option("inferSchema", "true")
             .csv(path)
    )
    df.write.mode("overwrite").format("delta").saveAsTable(table)

    print(f"{item['file']:<20} -> {table:<40} {df.count():>7} rows")
```
