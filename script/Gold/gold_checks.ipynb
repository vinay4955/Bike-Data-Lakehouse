```python
CATALOG = "workspace"
```

```python
def count(sql): return spark.sql(sql).collect()[0][0]
failures = []
def check(name, passed):
    print(("PASS" if passed else "FAIL"), "-", name)
    if not passed: failures.append(name)

silver_rows = count(f"SELECT COUNT(*) FROM {CATALOG}.silver.crm_sales")
gold_rows   = count(f"SELECT COUNT(*) FROM {CATALOG}.gold.fact_sales")
check("fact_sales row count matches silver", silver_rows == gold_rows)

silver_total = count(f"SELECT SUM(sales_amount) FROM {CATALOG}.silver.crm_sales")
gold_total   = count(f"SELECT SUM(sales_amount) FROM {CATALOG}.gold.fact_sales")
check("total sales amount matches silver", silver_total == gold_total)

assert not failures, f"Gold checks FAILED: {failures}"
print("Gold checks passed.")
```
