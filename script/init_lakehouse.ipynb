# Step 0: Create the Lakehouse
Run this notebook **once**. It creates the 3 layers (schemas) and the Volume where the raw CSV files live.

## Select the catalog

```python
%sql
USE CATALOG workspace;
```

```python
display(_sqldf)
```

## Create the 3 schemas (Bronze, Silver, Gold)

```python
%sql
CREATE SCHEMA IF NOT EXISTS bronze COMMENT 'Bronze layer: raw ingested data';
CREATE SCHEMA IF NOT EXISTS silver COMMENT 'Silver layer: cleaned and standardized data';
CREATE SCHEMA IF NOT EXISTS gold   COMMENT 'Gold layer: business-ready data';
```

```python
display(_sqldf)
```
